```python
# Unir orders e customers usando customer_id como chave
# how='left' mantém todos os pedidos, mesmo que algum cliente não seja encontrado

df = df_orders.merge(df_customers, on='customer_id', how='left')
```

```python
df.shape
```

```python
# Adicionar itens de cada pedido usando order_id como chave
df = df.merge(df_items, on='order_id', how='left')
```

```python
df.shape
```

```python
# Adicionar pagamentos
df = df.merge(df_payments, on='order_id', how='left')
```

```python
df.shape
```

```python
# Adicionar reviews
df = df.merge(df_reviews, on='order_id', how='left')
```

```python
df.shape
```

```python
# Adicionar produtos
df = df.merge(df_products, on='product_id', how='left')
```

```python
df.shape
```

```python
# Adicionar vendedores
df = df.merge(df_sellers, on='seller_id', how='left')
```

```python
df.shape
```

```python
df = df.merge(df_translation, on='product_category_name', how='left')
```

```python
df.shape
```

```python
# Converter colunas de data de texto para datetime
colunas_data = ['order_purchase_timestamp', 'order_approved_at', 
                 'order_delivered_carrier_date', 'order_delivered_customer_date', 
                 'order_estimated_delivery_date']

for col in colunas_data:
    df[col] = pd.to_datetime(df[col])
```

```python
df.info()
```

```python
# Tempo de entrega em dias (da compra até a entrega ao cliente)
df['tempo_entrega_dias'] = (df['order_delivered_customer_date'] - df['order_purchase_timestamp']).dt.days

# Atraso em dias (positivo = atrasou, negativo = chegou antes)
df['atraso_dias'] = (df['order_delivered_customer_date'] - df['order_estimated_delivery_date']).dt.days
```

```python
df[['tempo_entrega_dias', 'atraso_dias']].describe()
```

```python
#Pergunta 1: o prazo de entrega afeta a satisfação do cliente?
# Média de review_score por faixa de atraso (atrasado vs no prazo)
df['atrasado'] = df['atraso_dias'] > 0
df.groupby('atrasado')['review_score'].mean()
```

```python
#Pedidos entregues no prazo: nota média 4.13 (de 5)
#Pedidos atrasados: nota média 2.25
```

```python
#Utilizar o gráfico para melhor visualização
df.groupby('atrasado')['review_score'].mean().plot(kind='bar', color=['green', 'red'])
plt.title('Nota média por situação de entrega')
plt.xlabel('Atrasado')
plt.ylabel('Nota média (review_score)')
plt.xticks([0, 1], ['No prazo', 'Atrasado'], rotation=0)
plt.savefig('nota_por_atraso.png', bbox_inches='tight')
plt.show()
```

```python
#Pergunta 2: Quais categorias de produto têm mais atraso na entrega?

# Percentual de pedidos atrasados por categoria (top 10)
df.groupby('product_category_name_english')['atrasado'].mean().sort_values(ascending=False).head(10) * 100
```

```python
top_atraso = df.groupby('product_category_name_english')['atrasado'].mean().sort_values(ascending=False).head(10) * 100

top_atraso.plot(kind='barh', color='orange')
plt.title('Top 10 categorias com maior % de pedidos atrasados')
plt.xlabel('% de pedidos atrasados')
plt.ylabel('Categoria')
plt.gca().invert_yaxis()
plt.savefig('atraso_por_categoria.png', bbox_inches='tight')
plt.show()
```

```python
#Pergunta 3: quais estados têm os piores tempos de entrega?
# Percentual de pedidos atrasados por estado (top 10 piores)
df.groupby('customer_state')['atrasado'].mean().sort_values(ascending=False).head(10) * 100
```

```python
top_atraso_estado = df.groupby('customer_state')['atrasado'].mean().sort_values(ascending=False).head(10) * 100

top_atraso_estado.plot(kind='barh', color='crimson')
plt.title('Top 10 estados com maior % de pedidos atrasados')
plt.xlabel('% de pedidos atrasados')
plt.ylabel('Estado')
plt.gca().invert_yaxis()
plt.savefig('atraso_por_estado.png', bbox_inches='tight')
plt.show()
```
